```python
!pip install scikit-fuzzy

import numpy as np
import skfuzzy as fuzz
from skfuzzy import control as ctrl

order_density = ctrl.Antecedent(np.arange(0, 101, 1), 'order_density')
delivery_urgency = ctrl.Antecedent(np.arange(0, 101, 1), 'delivery_urgency')
driver_load = ctrl.Antecedent(np.arange(0, 101, 1), 'driver_load')
traffic_conditions = ctrl.Antecedent(np.arange(0, 101, 1), 'traffic_conditions')
profit_per_delivery = ctrl.Antecedent(np.arange(0, 101, 1), 'profit_per_delivery')

orders_to_combine = ctrl.Consequent(np.arange(0, 11, 1), 'orders_to_combine')
delivery_priority = ctrl.Consequent(np.arange(0, 101, 1), 'delivery_priority')

order_density['L'] = fuzz.trimf(order_density.universe, [0, 0, 40])
order_density['M'] = fuzz.trimf(order_density.universe, [30, 50, 70])
order_density['H'] = fuzz.trimf(order_density.universe, [60, 100, 100])

delivery_urgency['L'] = fuzz.trimf(delivery_urgency.universe, [0, 0, 40])
delivery_urgency['M'] = fuzz.trimf(delivery_urgency.universe, [30, 50, 70])
delivery_urgency['H'] = fuzz.trimf(delivery_urgency.universe, [60, 100, 100])

driver_load['L'] = fuzz.trimf(driver_load.universe, [0, 0, 40])
driver_load['M'] = fuzz.trimf(driver_load.universe, [30, 50, 70])
driver_load['H'] = fuzz.trimf(driver_load.universe, [60, 100, 100])

traffic_conditions['L'] = fuzz.trimf(traffic_conditions.universe, [0, 0, 40])
traffic_conditions['M'] = fuzz.trimf(traffic_conditions.universe, [30, 50, 70])
traffic_conditions['H'] = fuzz.trimf(traffic_conditions.universe, [60, 100, 100])

profit_per_delivery['L'] = fuzz.trimf(profit_per_delivery.universe, [0, 0, 40])
profit_per_delivery['M'] = fuzz.trimf(profit_per_delivery.universe, [30, 50, 70])
profit_per_delivery['H'] = fuzz.trimf(profit_per_delivery.universe, [60, 100, 100])

orders_to_combine['F'] = fuzz.trimf(orders_to_combine.universe, [0, 0, 3])
orders_to_combine['M'] = fuzz.trimf(orders_to_combine.universe, [2, 5, 7])
orders_to_combine['N'] = fuzz.trimf(orders_to_combine.universe, [6, 10, 10])

delivery_priority['L'] = fuzz.trimf(delivery_priority.universe, [0, 0, 40])
delivery_priority['M'] = fuzz.trimf(delivery_priority.universe, [30, 50, 70])
delivery_priority['H'] = fuzz.trimf(delivery_priority.universe, [60, 100, 100])

rule1 = ctrl.Rule(order_density['H'] & driver_load['L'] & traffic_conditions['L'], orders_to_combine['N'])
rule2 = ctrl.Rule(order_density['M'] & traffic_conditions['H'] & delivery_urgency['M'], orders_to_combine['M'])
rule3 = ctrl.Rule(driver_load['H'] & order_density['H'] & profit_per_delivery['M'], orders_to_combine['M'])
rule4 = ctrl.Rule(order_density['L'] & delivery_urgency['H'] & traffic_conditions['M'], orders_to_combine['M'])
rule5 = ctrl.Rule(profit_per_delivery['H'] & delivery_urgency['H'] & traffic_conditions['H'], orders_to_combine['M'])

rule6 = ctrl.Rule(delivery_urgency['H'] & profit_per_delivery['H'], delivery_priority['H'])
rule7 = ctrl.Rule(delivery_urgency['M'] & traffic_conditions['M'], delivery_priority['M'])
rule8 = ctrl.Rule(delivery_urgency['L'] & order_density['H'] & profit_per_delivery['L'], delivery_priority['L'])

rules = [rule1, rule2, rule3, rule4, rule5, rule6, rule7, rule8]

system_ctrl = ctrl.ControlSystem(rules)
system = ctrl.ControlSystemSimulation(system_ctrl)

system.input['order_density'] = 80
system.input['delivery_urgency'] = 50
system.input['driver_load'] = 30
system.input['traffic_conditions'] = 50
system.input['profit_per_delivery'] = 50

system.compute()

print("Số lượng đơn hàng cần kết hợp:", system.output['orders_to_combine'])
print("Mức độ ưu tiên giao hàng:", system.output['delivery_priority'])
```
