```python
!pip install scikit-fuzzy

import numpy as np
import skfuzzy as fuzz
from skfuzzy import control as ctrl

store_rating = ctrl.Antecedent(np.arange(0, 5.1, 0.1), 'store_rating')
sales_volume = ctrl.Antecedent(np.arange(0, 101, 1), 'sales_volume')
profit_margin = ctrl.Antecedent(np.arange(0, 101, 1), 'profit_margin')
seasonal_event = ctrl.Antecedent(np.arange(0, 101, 1), 'seasonal_event')
competitor_discount = ctrl.Antecedent(np.arange(0, 101, 1), 'competitor_discount')

discount = ctrl.Consequent(np.arange(0, 70.1, 0.1), 'discount')

store_rating['L'] = fuzz.trimf(store_rating.universe, [0, 0, 4.0])
store_rating['M'] = fuzz.trimf(store_rating.universe, [4.0, 4.25, 4.5])
store_rating['H'] = fuzz.trimf(store_rating.universe, [4.5, 5.0, 5.0])

sales_volume['L'] = fuzz.trimf(sales_volume.universe, [0, 0, 40])
sales_volume['M'] = fuzz.trimf(sales_volume.universe, [30, 50, 70])
sales_volume['H'] = fuzz.trimf(sales_volume.universe, [60, 100, 100])

profit_margin['L'] = fuzz.trimf(profit_margin.universe, [0, 0, 30])
profit_margin['M'] = fuzz.trimf(profit_margin.universe, [20, 50, 70])
profit_margin['H'] = fuzz.trimf(profit_margin.universe, [60, 100, 100])

seasonal_event['N'] = fuzz.trimf(seasonal_event.universe, [0, 0, 30])
seasonal_event['M'] = fuzz.trimf(seasonal_event.universe, [20, 50, 70])
seasonal_event['H'] = fuzz.trimf(seasonal_event.universe, [60, 100, 100])

competitor_discount['L'] = fuzz.trimf(competitor_discount.universe, [0, 0, 30])
competitor_discount['M'] = fuzz.trimf(competitor_discount.universe, [20, 50, 70])
competitor_discount['H'] = fuzz.trimf(competitor_discount.universe, [60, 100, 100])

discount['VL'] = fuzz.trimf(discount.universe, [0, 0, 5])       # Rất thấp
discount['L']  = fuzz.trimf(discount.universe, [5, 7.5, 10])    # Thấp
discount['M']  = fuzz.trimf(discount.universe, [10, 15, 20])    # Trung bình
discount['H']  = fuzz.trimf(discount.universe, [20, 30, 40])    # Cao
discount['VH'] = fuzz.trimf(discount.universe, [40, 55, 70])    # Rất cao

rule1 = ctrl.Rule(store_rating['H'] & sales_volume['H'] & profit_margin['H'], discount['VL'])
rule2 = ctrl.Rule(store_rating['L'] & sales_volume['L'] & profit_margin['H'], discount['H'])
rule3 = ctrl.Rule(seasonal_event['H'] & competitor_discount['H'], discount['VH'])
rule4 = ctrl.Rule(store_rating['M'] & sales_volume['M'] & profit_margin['M'], discount['M'])
rule5 = ctrl.Rule(competitor_discount['L'] & profit_margin['L'] & sales_volume['H'], discount['VL'])
rule6 = ctrl.Rule(store_rating['L'] & seasonal_event['N'], discount['M'])
rule7 = ctrl.Rule(sales_volume['L'] & profit_margin['L'], discount['VH'])

rules = [rule1, rule2, rule3, rule4, rule5, rule6, rule7]

discount_ctrl = ctrl.ControlSystem(rules)
discount_system = ctrl.ControlSystemSimulation(discount_ctrl)

discount_system.input['store_rating'] = 4.3
discount_system.input['sales_volume'] = 50
discount_system.input['profit_margin'] = 25
discount_system.input['seasonal_event'] = 80
discount_system.input['competitor_discount'] = 80

discount_system.compute()

print("Mức chiết khấu đề xuất:", discount_system.output['discount'])
```
