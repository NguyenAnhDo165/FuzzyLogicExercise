```python
!pip install scikit-fuzzy

iimport numpy as np
import skfuzzy as fuzz
from skfuzzy import control as ctrl

product_demand = ctrl.Antecedent(np.arange(0, 101, 1), 'product_demand')
competitor_pressure = ctrl.Antecedent(np.arange(0, 101, 1), 'competitor_pressure')
store_reputation = ctrl.Antecedent(np.arange(0, 5.1, 0.1), 'store_reputation')
profit_margin = ctrl.Antecedent(np.arange(0, 101, 1), 'profit_margin')
seasonal_demand = ctrl.Antecedent(np.arange(0, 101, 1), 'seasonal_demand')

discount = ctrl.Consequent(np.arange(0, 70.1, 0.1), 'discount')

product_demand['L'] = fuzz.trimf(product_demand.universe, [0, 0, 40])
product_demand['M'] = fuzz.trimf(product_demand.universe, [30, 50, 70])
product_demand['H'] = fuzz.trimf(product_demand.universe, [60, 100, 100])

competitor_pressure['L'] = fuzz.trimf(competitor_pressure.universe, [0, 0, 40])
competitor_pressure['M'] = fuzz.trimf(competitor_pressure.universe, [30, 50, 70])
competitor_pressure['H'] = fuzz.trimf(competitor_pressure.universe, [60, 100, 100])

store_reputation['L'] = fuzz.trimf(store_reputation.universe, [0, 0, 4.0])
store_reputation['M'] = fuzz.trimf(store_reputation.universe, [4.0, 4.25, 4.5])
store_reputation['H'] = fuzz.trimf(store_reputation.universe, [4.5, 5.0, 5.0])

profit_margin['L'] = fuzz.trimf(profit_margin.universe, [0, 0, 40])
profit_margin['M'] = fuzz.trimf(profit_margin.universe, [30, 50, 70])
profit_margin['H'] = fuzz.trimf(profit_margin.universe, [60, 100, 100])

seasonal_demand['N'] = fuzz.trimf(seasonal_demand.universe, [0, 0, 30])
seasonal_demand['M'] = fuzz.trimf(seasonal_demand.universe, [20, 50, 70])
seasonal_demand['H'] = fuzz.trimf(seasonal_demand.universe, [60, 100, 100])

discount['VL'] = fuzz.trimf(discount.universe, [0, 0, 5])
discount['L']  = fuzz.trimf(discount.universe, [5, 7.5, 10])
discount['M']  = fuzz.trimf(discount.universe, [10, 15, 25])
discount['H']  = fuzz.trimf(discount.universe, [20, 30, 40])
discount['VH'] = fuzz.trimf(discount.universe, [40, 55, 70])

rule1 = ctrl.Rule(product_demand['H'] & competitor_pressure['L'] & profit_margin['L'], discount['VL'])
rule2 = ctrl.Rule(product_demand['L'] & competitor_pressure['H'] & profit_margin['H'], discount['H'])
rule3 = ctrl.Rule(store_reputation['H'] & profit_margin['M'] & seasonal_demand['H'], discount['M'])
rule4 = ctrl.Rule(competitor_pressure['H'] & seasonal_demand['H'] & profit_margin['H'], discount['VH'])
rule5 = ctrl.Rule(store_reputation['L'] & product_demand['M'] & profit_margin['L'], discount['M'])
rule6 = ctrl.Rule(product_demand['H'] & seasonal_demand['N'] & competitor_pressure['L'], discount['VL'])
rule7 = ctrl.Rule(profit_margin['H'] & competitor_pressure['M'] & seasonal_demand['M'], discount['M'])

rules = [rule1, rule2, rule3, rule4, rule5, rule6, rule7]

system_ctrl = ctrl.ControlSystem(rules)
system = ctrl.ControlSystemSimulation(system_ctrl)
```
