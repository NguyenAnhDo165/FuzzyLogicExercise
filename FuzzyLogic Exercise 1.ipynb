```python
!pip install scikit-fuzzy

import numpy as np
import skfuzzy as fuzz
from skfuzzy import control as ctrl

ridedistance = ctrl.Antecedent(np.arange(0, 50.1, 0.1), 'ridedistance')
tfcond = ctrl.Antecedent(np.arange(0, 1.01, 0.01), 'tfcond')
demandlvl = ctrl.Antecedent(np.arange(0, 1.01, 0.01), 'demandlvl')
wthcond = ctrl.Antecedent(np.arange(0, 3.01, 0.01), 'wthcond')
rating = ctrl.Antecedent(np.arange(1.0, 5.1, 0.1), 'rating')
punct = ctrl.Antecedent(np.arange(0, 1.01, 0.01), 'punct')

price = ctrl.Consequent(np.arange(0, 4.01, 0.01), 'price')
bonus = ctrl.Consequent(np.arange(0, 4.01, 0.01), 'bonus')


ridedistance['S'] = fuzz.trimf(ridedistance.universe, [0, 0, 3])
ridedistance['M'] = fuzz.trimf(ridedistance.universe, [0, 2, 8])
ridedistance['L'] = fuzz.trimf(ridedistance.universe, [2, 6, 20])
ridedistance['VL'] = fuzz.trimf(ridedistance.universe, [15, 30, 50])

tfcond['L'] = fuzz.trimf(tfcond.universe, [0, 0, 0.3])
tfcond['M'] = fuzz.trimf(tfcond.universe, [0.2, 0.5, 0.7])
tfcond['H'] = fuzz.trimf(tfcond.universe, [0.6, 1, 1])


demandlvl['L'] = fuzz.trimf(demandlvl.universe, [0, 0, 0.3])
demandlvl['M'] = fuzz.trimf(demandlvl.universe, [0.2, 0.5, 0.7])
demandlvl['H'] = fuzz.trimf(demandlvl.universe, [0.6, 1, 1])


wthcond['G'] = fuzz.trimf(wthcond.universe, [0, 0, 1])
wthcond['M'] = fuzz.trimf(wthcond.universe, [0, 1.5, 2])
wthcond['B'] = fuzz.trimf(wthcond.universe, [1.5, 3, 3])


rating['P'] = fuzz.trimf(rating.universe, [1.0, 1.0, 2.5])
rating['A'] = fuzz.trimf(rating.universe, [2.0, 3.0, 4.0])
rating['G'] = fuzz.trimf(rating.universe, [3.5, 5.0, 5.0])


punct['L'] = fuzz.trimf(punct.universe, [0, 0, 0.5])
punct['OT'] = fuzz.trimf(punct.universe, [0.4, 0.6, 0.8])
punct['E'] = fuzz.trimf(punct.universe, [0.7, 1, 1])

price['L'] = fuzz.trimf(price.universe, [0, 0, 1])
price['M'] = fuzz.trimf(price.universe, [0.5, 1.5, 2.5])
price['H'] = fuzz.trimf(price.universe, [2, 3, 3.5])
price['VH'] = fuzz.trimf(price.universe, [3, 4, 4])

bonus['N'] = fuzz.trimf(bonus.universe, [0, 0, 1])
bonus['F'] = fuzz.trimf(bonus.universe, [0.5, 1.5, 2.5])
bonus['M'] = fuzz.trimf(bonus.universe, [2, 2.5, 3])
bonus['H'] = fuzz.trimf(bonus.universe, [3, 4, 4])

rule1  = ctrl.Rule(ridedistance['S']  & tfcond['L'] & demandlvl['L'], price['L'])
rule2  = ctrl.Rule(ridedistance['S']  & tfcond['M'] & demandlvl['H'], price['M'])
rule3  = ctrl.Rule(ridedistance['M']  & tfcond['H'] & demandlvl['H'], price['H'])
rule4  = ctrl.Rule(ridedistance['L']  & tfcond['M'] & wthcond['G'], price['M'])
rule5  = ctrl.Rule(ridedistance['L']  & tfcond['H'] & wthcond['B'], price['VH'])
rule6  = ctrl.Rule(ridedistance['VL'] & tfcond['H'] & demandlvl['H'], price['VH'])
rule7  = ctrl.Rule(ridedistance['M']  & tfcond['L'] & demandlvl['L'], price['M'])
rule8  = ctrl.Rule(ridedistance['S']  & tfcond['H'] & wthcond['B'], price['H'])
rule9  = ctrl.Rule(ridedistance['VL'] & wthcond['B'], price['VH'])
rule10 = ctrl.Rule(ridedistance['M']  & tfcond['M'] & wthcond['M'],price['M'])

rule11 = ctrl.Rule(rating['G'] & punct['E'], bonus['H'])
rule12 = ctrl.Rule(rating['A'] & punct['OT'], bonus['M'])
rule13 = ctrl.Rule(rating['P'] & punct['L'], bonus['N'])
rule14 = ctrl.Rule(ridedistance['L'] & tfcond['H'] & punct['OT'], bonus['H'])
rule15 = ctrl.Rule(ridedistance['M'] & tfcond['M'] & rating['G'], bonus['M'])
rule16 = ctrl.Rule(rating['P'] & punct['L'], bonus['N'])
rule17 = ctrl.Rule(ridedistance['VL'] & wthcond['B'] & rating['G'], bonus['H'])
rule18 = ctrl.Rule(ridedistance['S'] & rating['A'] & punct['OT'], bonus['F'])
rule19 = ctrl.Rule(ridedistance['L'] & tfcond['H'] & punct['L'], bonus['F'])
rule20 = ctrl.Rule(ridedistance['M'] & wthcond['M'] & rating['G'],bonus['M'])

rules = [rule1, rule2, rule3, rule4, rule5,
    rule6, rule7, rule8, rule9, rule10,
    rule11, rule12, rule13, rule14, rule15,
    rule16, rule17, rule18, rule19, rule20]

system_ctrl = ctrl.ControlSystem(rules)
system = ctrl.ControlSystemSimulation(system_ctrl)

system.input['ridedistance'] = float(input("Vui lòng nhập chiều dài (từ 1 đến 50 km): "))
system.input['tfcond'] = float(input("Vui lòng nhập tỉ lệ hoàn thành đơn hàng (từ 0 đến 1): "))
system.input['demandlvl'] = float(input("Vui lòng nhập mức độ nhu cầu (từ 0 đến 1): "))
system.input['wthcond'] = float(input("Vui lòng nhập tình trạng thời tiết (từ 0 đến 3): "))
system.input['rating'] = float(input("Vui lòng nhập xếp hạng (từ 1 đến 5): "))
system.input['punct'] = float(input("Vui lòng nhập độ phạt (từ 0 đến 1): "))
system.compute()
price.view(sim=system)
bonus.view(sim=system)
print(f"Thời gian giặt là: {system.output['price']:}")
print(f"Mực nước là: {system.output['bonus']:}")
```
